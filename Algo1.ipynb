```python

import random

n = int(input("Enter number of features: "))
population_size = int(input("Enter population size: "))
generations = int(input("Enter number of generations: "))
mutation_rate = float(input("Enter mutation rate (0 to 1): "))

def fitness(individual):
    return sum(individual)

population = []

for i in range(population_size):
    individual = [random.randint(0, 1) for j in range(n)]

    if sum(individual) == 0:
        individual[random.randint(0, n - 1)] = 1

    population.append(individual)

print("\nInitial Random Population:")
for i in range(population_size):
    print(population[i], "Fitness:", fitness(population[i]))

for generation in range(generations):
    population.sort(key=fitness, reverse=True)

    print("\nGeneration", generation + 1)
    print("Selected parents:", population[0], population[1]
          if population_size > 1 else population[0])

    new_population = [population[0][:]]

    while len(new_population) < population_size:
        p1 = population[0]
        p2 = population[1] if population_size > 1 else population[0]

        if n > 1:
            point = random.randint(1, n - 1)
            child = p1[:point] + p2[point:]
        else:
            child = p1[:]

        for i in range(n):
            if random.random() < mutation_rate:
                child[i] = 1 - child[i]

        if sum(child) == 0:
            child[random.randint(0, n - 1)] = 1

        new_population.append(child)

    population = new_population

    print("New population:")
    for individual in population:
        print(individual, "Fitness:", fitness(individual))

best = max(population, key=fitness)

print("\nBest feature subset:", best)
print("Selected feature numbers:",
      [i + 1 for i in range(n) if best[i] == 1])
print("Number of selected features:", sum(best))
```
