```python
import random

def booth(x, y):
    return (x + 2*y - 7)**2 + (2*x + y - 5)**2

num_particles = int(input("Enter number of particles: "))
iterations = int(input("Enter number of iterations: "))
w = float(input("Enter inertia weight (w): "))
c1 = float(input("Enter cognitive coefficient (c1): "))
c2 = float(input("Enter social coefficient (c2): "))

particles = []

for i in range(num_particles):
    x = random.uniform(-10, 10)
    y = random.uniform(-10, 10)
    vx = random.uniform(-1, 1)
    vy = random.uniform(-1, 1)

    particles.append({
        "position": [x, y],
        "velocity": [vx, vy],
        "best_position": [x, y],
        "best_value": booth(x, y)
    })

global_best_position = particles[0]["position"][:]
global_best_value = particles[0]["best_value"]

for particle in particles:
    if particle["best_value"] < global_best_value:
        global_best_value = particle["best_value"]
        global_best_position = particle["best_position"][:]

for iteration in range(iterations):
    for particle in particles:
        x, y = particle["position"]
        vx, vy = particle["velocity"]
        px, py = particle["best_position"]
        gx, gy = global_best_position

        r1 = random.random()
        r2 = random.random()

        vx = w * vx + c1 * r1 * (px - x) + c2 * r2 * (gx - x)
        vy = w * vy + c1 * r1 * (py - y) + c2 * r2 * (gy - y)

        x += vx
        y += vy

        particle["velocity"] = [vx, vy]
        particle["position"] = [x, y]

        value = booth(x, y)

        if value > particle["best_value"]:
            particle["best_value"] = value
            particle["best_position"] = [x, y]

        if value > global_best_value:
            global_best_value = value
            global_best_position = [x, y]

print("\nOptimal x:", global_best_position[0])
print("Optimal y:", global_best_position[1])
print("Minimum Booth function value:", global_best_value)
```
